```python
import sys
from pathlib import Path

project_root = Path.cwd().parent    # notebooks -> project_root
sys.path.insert(0, str(project_root))
print(project_root)

import numpy as np
from FenicsxModels import FenicsxPriors, Helpers
from dolfinx import mesh, fem
from mpi4py import MPI
```

```python
# mesh size
Nx = 64
Ny = 64
corner_points=((0.0, 0.0), (2.0, 2.0))

msh = mesh.create_rectangle(
    comm=MPI.COMM_WORLD,
    points=corner_points,
    n=(Nx, Ny),
    cell_type=mesh.CellType.triangle,
)

Kh = fem.functionspace(msh, ("DG", 0))

# one parameter per field
n_fields = 3
ell = (0.05, 0.1, 0.2)
sigma=(0.5, 0.7, 1.0)
mu=(0.0, 0.5, -0.5)
rho = 0.9
corr = (1.0 - rho) * np.eye(n_fields) + rho * np.ones((n_fields, n_fields))

prior = FenicsxPriors.MultiFieldPrior(
    Kh,
    sigma=sigma,
    ell=ell,
    mu=mu,
    correlation=corr,
    names=["field_0", "field_1", "field_2"],
    seed=4000,
)

k = prior.sample()
Helpers.plot_prior_draw_mpi(
    prior,
    k,
    filename=None,
    show=True,
)
```
