```python
import sympy as sp

# variables
x, y = sp.symbols('x y', real=True)

# parameters
alpha = 0.5
cL, cR = 1.0, 0.2
bL, bR = 0.1, -0.05
Aamp = 0.3

# permeability
k = 1 + alpha * sp.sin(2*sp.pi*x) * sp.cos(2*sp.pi*y)

# boundary data
gL = cL + bL * sp.cos(2*sp.pi*y)
gR = cR + bR * sp.cos(2*sp.pi*y)

# exact solution
p_ex = (1 - x)*gL + x*gR + Aamp * sp.sin(sp.pi*x) * sp.cos(2*sp.pi*y)

# gradient
grad_px = sp.diff(p_ex, x)
grad_py = sp.diff(p_ex, y)

# flux
qx = k * grad_px
qy = k * grad_py

# divergence
div_q = sp.diff(qx, x) + sp.diff(qy, y)

# forcing
f = -sp.simplify(div_q)

print("f(x,y) =")
sp.pprint(f)

bc_left = sp.simplify(p_ex.subs(x,0) - gL)
print(bc_left)

bc_right = sp.simplify(p_ex.subs(x,1) - gR)
print(bc_right)

bc_top = sp.simplify(qy.subs(y,0))
print(bc_top)

bc_bottom = sp.simplify(qy.subs(y,1))
print(bc_bottom)
```
